```python
import pandas as pd
```

```python
# =========================================================
# LOAD DATASET
# =========================================================

print("Loading dataset...")

data = pd.read_csv('./AIML Dataset.csv')

print("The dataset was successfully read..")
print(f"Number of data: {len(data):,}")

```

```python
# =========================================================
# DISTRIBUSI CHECK
# =========================================================

print("\nInitial distribution:")

print(
    data["isFraud"].value_counts()
)

print(
    data["isFraud"].value_counts(normalize=True) * 100
)
```

```python
# =========================================================
# STRATIFIED SAMPLING
# =========================================================

# Desired sample size
SAMPLE_SIZE = 100_000

# The proportion of fraud and non-fraud cases is maintained.
sample_data = (
    data
    .groupby("isFraud", group_keys=False)
    .apply(
        lambda x: x.sample(
            n=min(
                len(x),
                round(
                    SAMPLE_SIZE
                    * len(x)
                    / len(data)
                )
            ),
            random_state=42
        )
    )
)
```

```python
# =========================================================
# SHUFFLE DATA
# =========================================================

sample_data = sample_data.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
# =========================================================
# SAVE SAMPLE
# =========================================================

sample_data.to_csv(
    "fraud_sample.csv",
    index=False
)
```

```python
# =========================================================
# HASIL
# =========================================================

print("\n===================================")
print("Sampling selesai!")
print("===================================")

print(
    f"Jumlah sample: {len(sample_data):,}"
)

print(
    f"Ukuran file: {sample_data.memory_usage(deep=True).sum() / 1024**2:.2f} MB"
)

print("\nDistribusi sample:")

print(
    sample_data["isFraud"].value_counts()
)

print("\nProporsi sample:")

print(
    sample_data["isFraud"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nFile tersimpan sebagai:")
print("fraud_sample.csv")
```
